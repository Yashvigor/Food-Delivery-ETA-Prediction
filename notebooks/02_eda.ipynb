# Data Science Lifecycle: Part 2 - Exploratory Data Analysis (EDA)

Understanding our logistics features, targets, and distributions through visual graphics and statistical metrics.

### Objectives:
1. **Univariate Analysis**: Inspect distribution shapes of distance, prep times, and target delivery times.
2. **Bivariate Analysis**: Review how distance, courier experience, and traffic severity impact the target variable.
3. **Multivariate Analysis**: Observe compound weather, traffic, and distance impacts on transit delays.
4. **Correlation Matrix Heatmap**: Quantify correlations among all continuous numerical metrics.
5. **Outlier Visualization**: Inspect outlier bands with boxplots.

---

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="darkgrid")
processed_csv_path = "../data/processed/processed_deliveries.csv"
df = pd.read_csv(processed_csv_path)
df.head()
```

## 1. Univariate Analysis

Checking feature ranges and frequencies.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Distance Distribution
sns.histplot(df["Distance_km"], kde=True, ax=axes[0], color="#4facfe")
axes[0].set_title("Distribution of Distance (km)")
axes[0].set_xlabel("Distance (km)")

# Delivery Time Distribution
sns.histplot(df["Delivery_Time_Min"], kde=True, ax=axes[1], color="#00f2fe")
axes[1].set_title("Distribution of Delivery Time (Min)")
axes[1].set_xlabel("Delivery Time (Minutes)")

# Preparation Time Distribution
sns.histplot(df["Preparation_Time"], kde=True, ax=axes[2], color="#ff9f43")
axes[2].set_title("Distribution of Preparation Time (Min)")
axes[2].set_xlabel("Preparation Time (Minutes)")

plt.tight_layout()
plt.show()
```

## 2. Bivariate Analysis

Analyzing pairwise relationships with target variable `Delivery_Time_Min`.

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Distance vs Delivery Time
sns.scatterplot(data=df, x="Distance_km", y="Delivery_Time_Min", alpha=0.3, ax=axes[0], color="#4facfe")
axes[0].set_title("Distance (km) vs Delivery Time (Min)")

# 2. Courier Experience vs Delivery Time
sns.boxplot(data=df, x="Courier_Experience", y="Delivery_Time_Min", ax=axes[1], palette="coolwarm")
axes[1].set_title("Courier Experience (Years) vs Delivery Time (Min)")

# 3. Traffic Level vs Delivery Time
sns.boxplot(data=df, x="Traffic_Level", y="Delivery_Time_Min", order=["Low", "Medium", "High", "Jam"], ax=axes[2], palette="autumn")
axes[2].set_title("Traffic Level vs Delivery Time (Min)")

plt.tight_layout()
plt.show()
```

## 3. Multivariate Analysis

Compounding features to observe transit variance.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df, 
    x="Distance_km", 
    y="Delivery_Time_Min", 
    hue="Traffic_Level", 
    style="Weather", 
    palette="viridis",
    alpha=0.6
)
plt.title("Compound Impact: Traffic, Weather, and Distance vs Delivery Minutes")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

## 4. Correlation Analysis

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
corr_matrix = df[numeric_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5, vmin=-1.0, vmax=1.0)
plt.title("Correlation Heatmap of Continuous Logistics Metrics")
plt.show()
```
